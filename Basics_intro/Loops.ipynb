## Loop in python

```python
for i in range(10, 20):
    print(i)
```

```python
for i in range(3, -50, -5):
    print(i)
```

__with string__

```python
str="Beluga"
for i in str:
    print(i)
```

__while loop__

```python
count = -5

while count%2==0:
    print(count)

while (count%5==0) and (count>-10000):
    print(count)
    count = count*2
```

__Loop control statements : `break and continue`__

```python
for i in range(2,20):
    if i>=15:
        break
    if i%2==0:
        continue
    print(i)
```

__Nested loop__

```python
for i in range(5):
    for j in range(2):
        print(f"i : {i} and j : {j} = {i and j}")
```

## Calculate the sum of first natural num using loop

```python
n = int(input("Enter a natural number : "))

sum=0
intN=1

while n>=intN :
    sum += intN 
    print(f"The sum of first {intN} natural num is : {sum}")

    intN += 1 
```

```python
n = int(input("Enter a natural number : "))
sum = 0

for i in range(n+1) :
    sum += i

print(f"The sum of first {n} natural num is : {sum}")
```

## Find all prime num in range

```python
for num in range(2, 101):
    for i in range(2, num):
        
        if num % i == 0:
                break
        else : 
            print(num)
```
