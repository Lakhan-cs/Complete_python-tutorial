`packing and unpacking` __of tuple__

```python
packed_tuple = 2, 3.14, 'Dear User', True
print(packed_tuple)
```

__Unpacking__

```python
a, b, c, d = packed_tuple
print(a); print(b); print(c); print(d)
```

__Using *__

```python
num = (1,3,4,5,6,8,22,66,7,14)
i, *mid, j = num
print(i); print(mid); print(j)
```

__Nested tuple__

```python
lst = [(2,3,5,2), ['a', 'b', 'c'], {3, 3.14, True, False}]

print(lst)
print(lst[1][2])
print(lst[2])
```

__Iterating over nested__

```python
for sub_lst in lst:
    for item in sub_lst:
        print(item, end = ' ')
    print( )
```

## SETS

__Oparetion on set__

```python
my_set= {1,2,3,4,6,8,4,23,2,2,1,1,2,1,4,2,1,2}

my_set.pop()

print(my_set)
print(1 in my_set)
print(3 in my_set)
print(56 in my_set)
print(2 in my_set)
```

```python
set1 = {2, 44, 22, 11, 33, 2, 66, 00, 99}
set2 = {55, 1,3, 44, 55,77, 88, 1011, 4, 11, 2, 0}

I = set1.intersection(set2)
U = set1.union(set2)

print(f"set1 | set2 = {sorted(U)}")
print(f"set1 & set2 = {I}")

# common elements of set1 will be updated to set2
set1.intersection_update(set2)
print(set1); print(set2)
```

__Difference of sets__

```python
set1 = {2, 44, 22, 11, 33, 2, 66, 00, 99}
set2 = {55, 1,3, 44, 55,77, 88, 1011, 4, 11, 2, 0}

# only set1 elements
print(set1.difference(set2))
```

`A|B - A&B`

```python
set1.symmetric_difference(set2)
```

__Sets methods__

```python
# is subset
s1 = {1,2,4,3,4,5,7,6,8,9}
s2= {1,2,3}
print(s1.issubset(set2))

# is superset
print(s1.issuperset(s2))
```
