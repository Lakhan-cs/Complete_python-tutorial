### Why function ?

`We can make our own function for specific task`

__e.i. - In DS task have to make multiple functions for specific task__

    Data Preprocessing ... loading

    Training ... predicting

    Forecasting

```python
def even_odd(n):
    """Will create own function for a task... and then can call for another task"""
    
    if n%2==0:
        print(f"{n} is even num.")
    else :
        print(f"{n} is odd num.")
```

__Call a function__

```python
even_odd(-38)
even_odd(8)
even_odd(87)
even_odd(0.0)
even_odd(-2)
even_odd(-10) 
```

__function with multiple parameters__

```python
def percent(a, b, c, d, e):
    percent = ((a+b+c+d+e)*100)/500
    return percent

percent(88, 91, 87, 95, 94)
```

__Positional argument `(* used for multi parametes)`__

```python
def positional_args(*para):
    print("Thiss will help to not define all parameter they r :")
    for i in para:
        print(f" - {i}")


positional_args('Bello', 'string', True, 'bool', 3.14, 'numeric', ["Can", "add", "any, 'parameters"], "List ...")
```

__keyword args `(** used for dict form)`__

```python
def keyword_args(**dict):
    print("Thiss helps too in to not define all parameters of Dict")
    for key, value in dict.items():
        print(f"- {key} : {value}")

keyword_args(args='keyword', helps_in='Dict parameter', param_define=False) 
```

__Combined positinal n keyword arg__

```python
def multi_args(*args, **dict):
    print("Thiss will help to not define all parameter they r :")
    for i in args:
        print(f" - {i}")


    print("Thiss helps too in to not define all parameters of Dict")
    for key, value in dict.items():
        print(f"- {key} : {value}")
```

```python
multi_args("Positional args", 
           [3.14, True, 00], 
           {"rn":"right now", "wyd":"what r u doin", "fr":"for real"},
           "W":"win",
           "prolly":"probably",
           "mb":"my bad"
           )
```
