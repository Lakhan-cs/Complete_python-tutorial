## Packages

```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline 
import pandas as pd 
import seaborn as sns
```

__Load a sample dataset__

```python
cars = sns.load_dataset("mpg")
cars.head()
```

__Get itss statistics analysis__

```python
cars.describe()
```

__Find %ile of particular records__

```python
np.percentile(cars["mpg"], [25, 50, 75, 99, 100])
```

__boxploting__

```python
sns.boxplot(cars["mpg"])
plt.title("Boxplot of mpg")
plt.show()
```

__CHARTS n Visual Representation__

```python
sns.countplot(cars["origin"])  
plt.title("Charts of origin")
plt.show()
```

```python
sns.displot(cars["horsepower"]) 
plt.show()
```

__Customized countplot for 'mpg'__

```python
sns.countplot(x="mpg", data=cars, palette="Set2", edgecolor="black")
plt.xlabel("X-axis", fontsize=8)
plt.ylabel("Y-axis", fontsize=13)
plt.title("Charts", fontsize=20)
plt.tight_layout()
plt.show()
```

```python
sns.scatterplot(x="mpg", y="horsepower", data=cars)
plt.xlabel("miles per gain (mpg)")
plt.ylabel("horsepower")
plt.title("the comparition of two charts")
plt.show()
```
