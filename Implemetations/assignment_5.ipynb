#### Basic set operations

```python
import random 

set1 = set(random.randint(20, 50) for _ in range(10))
set2 = set(random.randint(30, 60) for _ in range(15))

print(f"A = {set1}\
       B = {set2}")

print(f"Union = {set1.union(set2)}")
print(f"Intersection = {set1.intersection(set2)}")
print(f"Difference = {set1.difference(set2)}")
print(f"U of !I = {set1.symmetric_difference(set2)}")
```

#### Find unique element of a random tuple

```python
import random
tpl = tuple(random.randint(1, 7) for _ in range(25))

print(f"Tuple = {sorted(tpl)}")
print(f"Unique Tuple = {tuple(set(tpl))}")
```

#### create a dict with set as keys

```python
game = {
    frozenset({"username", "bello.hello"}): 91,
    frozenset({"username", "urMaster"}): 75,
    frozenset({"username", "dotEXE"}): 87
}

for player, score in game.items():
    print(f"- {player} : {score}")
```

#### defualt dict

```python
from collections import defaultdict

default_d = defaultdict(list)
default_d["a"].append(1)
default_d["a"].append(2)
default_d["b"].append(3)
default_d["z"].append(0)

print(default_d)
```

#### counting char of a string

```python
def count_chars(s):
    count_dict = {}
    for char in s:
        count_dict[char] = count_dict.get(char, 0) + 1
    return count_dict

string = "Hello Dear User as part of learning python ASSESSMENT-V is Here."
print(count_chars(string))
```
